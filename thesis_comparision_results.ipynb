```python
#comparing ML and DL models

# ==========================================================
# CENTRALIZED vs DISTRIBUTED COMPARISON
# LR, DT, RF, ANN, GRU
# ==========================================================

%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np



comparison_data = {
    'Model': ['LR', 'DT', 'RF', 'ANN', 'GRU'],

    # Distributed Results
    'Distributed_Accuracy': [
        0.4160,   # LR
        0.6500,   # DT
        0.7240,   # RF
        0.6077,   # ANN (3M representative)
        0.9287    # GRU (5M representative)
    ],

    'Distributed_Precision': [
        0.5520,
        0.6600,
        0.7140,
        0.6375,
        0.9289
    ],

    'Distributed_Recall': [
        0.4160,
        0.6500,
        0.7240,
        0.6077,
        0.9287
    ],

    'Distributed_F1': [
        0.2550,
        0.6400,
        0.7040,
        0.6040,
        0.9282
    ],

    # Centralized Results
    'Centralized_Accuracy': [
        0.4500,   # Example placeholder
        0.7000,   # Example placeholder
        0.7800,   # Example placeholder
        0.9673,   # ANN centralized
        0.7091    # GRU centralized
    ],

    'Centralized_Precision': [
        0.5800,
        0.7100,
        0.7900,
        0.9679,
        0.7368
    ],

    'Centralized_Recall': [
        0.4500,
        0.7000,
        0.7800,
        0.9673,
        0.7091
    ],

    'Centralized_F1': [
        0.3000,
        0.6900,
        0.7700,
        0.9674,
        0.6247
    ]
}

df = pd.DataFrame(comparison_data)


def plot_metric(metric_name, distributed_col, centralized_col):
    x = np.arange(len(df['Model']))
    width = 0.35

    plt.figure(figsize=(12,6))

    plt.bar(x - width/2, df[distributed_col], width, label='Distributed')
    plt.bar(x + width/2, df[centralized_col], width, label='Centralized')

    plt.title(f'{metric_name}: Centralized vs Distributed Comparison')
    plt.xlabel('Models')
    plt.ylabel(metric_name)
    plt.xticks(x, df['Model'])
    plt.ylim(0, 1.05)
    plt.legend()
    plt.grid(axis='y')

    plt.tight_layout()
    plt.show()



plot_metric('Accuracy', 'Distributed_Accuracy', 'Centralized_Accuracy')

plot_metric('Precision', 'Distributed_Precision', 'Centralized_Precision')

plot_metric('Recall', 'Distributed_Recall', 'Centralized_Recall')

plot_metric('F1 Score', 'Distributed_F1', 'Centralized_F1')

```

```python
import matplotlib.pyplot as plt
import numpy as np

# ==============================
# Centralized Dataset All Models
# ==============================

models = [
    'Logistic Regression',
    'Decision Tree',
    'Random Forest',
    'ANN',
    'GRU'
]

accuracy = [0.6720, 0.9485, 0.9719, 0.9673, 0.7273]
precision = [0.7800, 0.9495, 0.9720, 0.9674, 0.7423]
recall = [0.6720, 0.9485, 0.9719, 0.9679, 0.7273]
f1_score = [0.6680, 0.9489, 0.9717, 0.9673, 0.6671]

# X-axis positions
x = np.arange(len(models))
width = 0.2

# ==============================
# Plot
# ==============================

plt.figure(figsize=(18,8))

bars1 = plt.bar(x - 1.5*width, accuracy, width, label='Accuracy')
bars2 = plt.bar(x - 0.5*width, precision, width, label='Precision')
bars3 = plt.bar(x + 0.5*width, recall, width, label='Recall')
bars4 = plt.bar(x + 1.5*width, f1_score, width, label='F1-Score')

# ==============================
# Labels and Formatting
# ==============================

plt.xticks(x, models, rotation=25, fontsize=11)
plt.ylabel("Performance Score", fontsize=12)
plt.xlabel("Models", fontsize=12)
plt.ylim(0, 1.05)
plt.title("Centralized Dataset Performance Comparison Across All Models", fontsize=15)
plt.legend()

# ==============================
# Add Value Labels
# ==============================

def add_labels(bars):
    for bar in bars:
        height = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width()/2,
            height + 0.008,
            f'{height:.3f}',
            ha='center',
            va='bottom',
            fontsize=8
        )

add_labels(bars1)
add_labels(bars2)
add_labels(bars3)
add_labels(bars4)

# ==============================
# Save High-Resolution Image
# ==============================

plt.tight_layout()
plt.savefig("centralized_all_models_performance.png", dpi=300)

# ==============================
# Show Graph
# ==============================

plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
import numpy as np

# Data
window_sizes = np.array([2, 3, 5, 7]).reshape(-1, 1)

metrics = np.array([
    [0.9976, 155.14, 219.81],
    [0.9945, 224.34, 277.07],
    [0.9881, 294.68, 349.13],
    [0.9835, 319.62, 394.99]
])

# Normalize
scaler = MinMaxScaler()
normalized_metrics = scaler.fit_transform(metrics)

# Split
correlation = normalized_metrics[:,0]
mad = normalized_metrics[:,1]
avg_std = normalized_metrics[:,2]

# Plot
plt.figure(figsize=(10,6))

plt.plot(window_sizes, correlation, marker='o', label='Correlation')
plt.plot(window_sizes, mad, marker='o', label='MAD')
plt.plot(window_sizes, avg_std, marker='o', label='AVG STD')

plt.axvline(x=3, linestyle='--', label='Selected Window Size = 3')

plt.xlabel("Window Size")
plt.ylabel("Normalized Metric Value")
plt.title("Normalized Window Size Optimization Metrics")
plt.xticks([2,3,5,7])
plt.legend()
plt.grid(True)

plt.show()
```
