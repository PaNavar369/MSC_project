```python
from pyspark.sql import SparkSession
spark= SparkSession.builder\
       .appName("Comparision")\
       .getOrCreate()
spark
       
  
```

```python
distributed_df=spark.read.parquet("/home/ubuntu/thesis/data/processed_data/system_state/distributed_data")
distributed_df.show(5)
#distributed_df.count()
```

```python
centralized_df= spark.read.parquet("/home/ubuntu/thesis/data/processed_data/system_state/centralized_data")
centralized_df.show(5)
```

```python
# compression ratio

D_count= distributed_df.count()
C_count= centralized_df.count()

compression_ratio= D_count/C_count

print("compression ratio is :" , compression_ratio)
```

```python
#The centralized representation comprises the data set approximately by 10⁵ times --- massive compression
# This leads the reduction of data granularity and eleiminating fine-grained behavioural variations

# so accuracy of centralized_dataset= 97%
# accuracy of distributed dataset=71%    due to this behavoiur data compression

# centralized_dataset----> simple and easy to learn
# distributed-dataset ----> complex( real-world difficulty)
```

```python
# variability ratio

from pyspark.sql.functions import stddev, avg

# Distributed
stats_D = distributed_df.select(
    stddev("value").alias("std"),
    avg("value").alias("mean")
).collect()[0]

cv_D = stats_D["std"] / stats_D["mean"]

# Centralized
stats_C = centralized_df.select(
    stddev("value").alias("std"),
    avg("value").alias("mean")
).collect()[0]

cv_C = stats_C["std"] / stats_C["mean"]

print("Distributed CV:", cv_D)
print("Centralized CV:", cv_C)


variability_ratio= cv_D/cv_C

print("variability_ratio:" , variability_ratio)
```

```python
# This disrtibuted dataset exhibits 300 times variety of data compared to centralized dataset(real world behaviour)
```

```python

import math

def compute_entropy(df):
    total = df.count()
    probs = df.groupBy("state_label").count().collect()
    
    entropy = 0
    for row in probs:
        p = row['count'] / total
        entropy -= p * math.log2(p)
    
    return entropy

entropy_D = compute_entropy(distributed_df)
entropy_C = compute_entropy(centralized_df)

print("Distributed entropy:", entropy_D)
print("Centralized entropy:", entropy_C)
```

```python
# calculation include in PPT
# This entropy is calculated to prove the class imbalaces in ML 
#  the reference point is -p*log(p)  where p is number of classes
```

```python
def redundancy_ratio(df, key_cols):
    df = df.repartition(200)
    df = df.cache()
    total = df.count()
    unique = df.select(key_cols).dropDuplicates().count()
    
    duplicates = total - unique
    ratio = duplicates / total
    
    return {
        "total": total,
        "unique": unique,
        "duplicates": duplicates,
        "redundancy_ratio": ratio
    }
```

```python
#Information loss
centralized_unique = centralized_df.select("value").distinct().count()
distributed_unique = distributed_df.select("value").distinct().count()

info_loss = 1 - (centralized_unique / distributed_unique)
print("Information Loss:", info_loss)
```

```python
#Centralized data achieves compact representation but at the cost of significant information loss and reduced variability. 
#Distributed data preserves fine-grained patterns and scales effectively,
#making it more suitable for real-world machine learning applications.”
```

```python
# Tredn distortion
from pyspark.sql.functions import variance

centralized_df.select(variance("trend")).show()
distributed_df.select(variance("trend")).show()
```

```python
#Distributed data exhibits significantly higher trend variance compared to centralized data, indicating that centralized aggregation smooths temporal dynamics and suppresses important fluctuations.”
```

```python
# Availability 

from pyspark.sql.functions import col
centralized_total = centralized_df.count()

centralized_normal = centralized_df.filter(col("state_label") == "Normal").count()
centralized_degraded = centralized_df.filter(col("state_label") == "Degraded").count()
centralized_failure = centralized_df.filter(col("state_label") == "Failure").count()

# Strict Availability (Normal only)
centralized_availability_strict = centralized_normal / centralized_total

print(centralized_availability_strict)
# Functional Availability (Normal + Degraded)
centralized_availability_functional = (
    centralized_normal + centralized_degraded
) / centralized_total

print(centralized_availability_functional)
centralized_reliability = 1 - (centralized_failure / centralized_total)
print(centralized_reliability)
```

```python

distributed_total = distributed_df.count()

distributed_normal = distributed_df.filter(col("state_label") == "Normal").count()
distributed_degraded = distributed_df.filter(col("state_label") == "Degraded").count()
distributed_failure = distributed_df.filter(col("state_label") == "Failure").count()

# Strict Availability
distributed_availability_strict = distributed_normal / distributed_total

print("distributed_strict_availability:", distributed_availability_strict)
# Functional Availability
distributed_availability_functional = (
    distributed_normal + distributed_degraded
) / distributed_total

print("distributed_availaibility_functional:",distributed_availability_functional)
# Reliability
distributed_reliability = 1 - (distributed_failure / distributed_total)

print("Reliability_distributed", distributed_reliability)
```
