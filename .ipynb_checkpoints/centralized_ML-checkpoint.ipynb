```python
from pyspark.sql import SparkSession
spark= SparkSession.builder\
       .appName("System_State_definitions")\
       .getOrCreate()
spark
```

```python
centralized_df= spark.read.parquet("/home/ubuntu/thesis/data/processed_data/system_state/centralized_data")
centralized_df.show(5)
```

```python
centralized_df.count()
```

```python
feature_columns=['value','Moving_avg','Moving_std','lag_1','lag_2','trend']
label='state_label'
```

```python
#normalizing the label into numbers
#0-Normal
#1 -degraded
#2 -failure
from pyspark.ml.feature import StringIndexer
indexer=StringIndexer(inputCol='state_label',outputCol='label')
C_df=indexer.fit(centralized_df).transform(centralized_df)
feature_columns=['value','Moving_avg','Moving_std','lag_1','lag_2','trend']

C_df.select(*feature_columns,'label')

C_df.show()
```

```python
C_df_new= C_df.drop("z_score")
C_df_new.show(5)
```

```python
from pyspark.sql.functions import col, isnan, when, count
C_df_new.select([
    count(when(col(c).isNull(), c)).alias(c)
    for c in C_df_new.columns
]).show()
```

```python
from pyspark.sql.functions import col
feature_columns=['value','Moving_avg','Moving_std','lag_1','lag_2','trend']
C_df_new=C_df_new.dropna(subset=feature_columns)
```

```python
# after removing the NUll valeus 

from pyspark.sql.functions import col, isnan, when, count
C_df_new.select([
    count(when(col(c).isNull(), c)).alias(c)
    for c in C_df_new.columns
]).show()
```

```python
C_df_clean=C_df_new
C_df_clean.show(5)
```

```python
C_df_clean.groupBy("label").count().show()
```

```python
from pyspark.ml.feature import VectorAssembler
feature_columns=['value','Moving_avg','Moving_std','lag_1','lag_2','trend']
assembler=VectorAssembler(
    inputCols=feature_columns,
    outputCol="features")


C_df_model=assembler.transform(C_df_clean)
C_df_model.count()

train = C_df_model.filter(col("Date") < "2016-06-01")
test  = C_df_model.filter(col("Date") >= "2016-06-01")

print("Train:", train.count())
print("Test:", test.count())
```

```python
# for Logistic regression
from pyspark.ml.classification import LogisticRegression
lr=LogisticRegression(
    featuresCol="features",
    labelCol="label",
    maxIter=30,
    regParam=0.1 
)

model=lr.fit(train)
```

```python
#predictions

predictions= model.transform(test)
predictions.select("features", "label", "prediction").show()
```

```python
# evaluation for accuracy

from pyspark.ml.evaluation import MulticlassClassificationEvaluator
evaluator=MulticlassClassificationEvaluator(
    labelCol="label",
    predictionCol="prediction",
    metricName="accuracy"
)
accuracy = evaluator.evaluate(predictions)
print("Accuracy for LR:", accuracy)

from pyspark.ml.evaluation import MulticlassClassificationEvaluator
evaluator= MulticlassClassificationEvaluator(
    labelCol="label",
    predictionCol="prediction",
    metricName="weightedRecall"
)
Recall=evaluator.evaluate(predictions)
print("Recall for LR:",Recall)

from pyspark.ml.evaluation import MulticlassClassificationEvaluator
evaluator= MulticlassClassificationEvaluator(
             labelCol="label",
             predictionCol="prediction",
            metricName="weightedPrecision"
)

precesion=evaluator.evaluate(predictions)
print("precesion for LR:", precesion)

from pyspark.ml.evaluation import MulticlassClassificationEvaluator
evaluator= MulticlassClassificationEvaluator(
            labelCol="label",
             predictionCol="prediction",
             metricName="f1"
)
f1= evaluator.evaluate(predictions)
print("F1 score for LR: ",f1)
```

```python
# evaluation for Recall
```

```python
# Random Forest 

from pyspark.ml.classification import RandomForestClassifier
from pyspark.ml.evaluation import MulticlassClassificationEvaluator

rf = RandomForestClassifier(
    featuresCol="features",
    labelCol="label",
    numTrees=50,
    maxDepth=5
)

from pyspark.ml.evaluation import MulticlassClassificationEvaluator
model = rf.fit(train)
predictions = model.transform(test)
```

```python
# ML metrics for RF

# Accuracy
evaluator = MulticlassClassificationEvaluator(
    labelCol="label",
    predictionCol="prediction",
    metricName="accuracy"
)
accuracy=evaluator.evaluate(predictions)
print("Accuracy for RF:",accuracy)

# Precesion

evaluator= MulticlassClassificationEvaluator(
             labelCol="label",
             predictionCol="prediction",
            metricName="weightedPrecision"
)

precesion=evaluator.evaluate(predictions)
print("precesion for RF:", precesion)

# Recall
evaluator= MulticlassClassificationEvaluator(
    labelCol="label",
    predictionCol="prediction",
    metricName="weightedRecall"
)
Recall=evaluator.evaluate(predictions)
print("Recall for RF:",Recall)

# F1 Score
evaluator_f1 = MulticlassClassificationEvaluator(
    labelCol="label",
    predictionCol="prediction",
    metricName="f1"
)
f1=evaluator_f1.evaluate(predictions)
print("F1 Score for RF:", f1)
```

```python
# Decision tree
from pyspark.ml.classification import DecisionTreeClassifier

dt = DecisionTreeClassifier(
    featuresCol="features",  
    labelCol="label",                
    maxDepth=5                      
)

model = dt.fit(train)
predictions = model.transform(test)
```

```python
# ML metrics for RF

# Accuracy
evaluator = MulticlassClassificationEvaluator(
    labelCol="label",
    predictionCol="prediction",
    metricName="accuracy"
)
accuracy=evaluator.evaluate(predictions)
print("Accuracy:",accuracy)

# Precesion

evaluator= MulticlassClassificationEvaluator(
             labelCol="label",
             predictionCol="prediction",
            metricName="weightedPrecision"
)

precesion=evaluator.evaluate(predictions)
print("precesion:", precesion)

# Recall
evaluator= MulticlassClassificationEvaluator(
    labelCol="label",
    predictionCol="prediction",
    metricName="weightedRecall"
)
Recall=evaluator.evaluate(predictions)
print("Recall",Recall)

# F1 Score
evaluator_f1 = MulticlassClassificationEvaluator(
    labelCol="label",
    predictionCol="prediction",
    metricName="f1"
)
f1=evaluator_f1.evaluate(predictions)
print("F1 Score:", f1)
```

```python
# for scalability  calculation


df_1 = C_df_model.sample(fraction=0.2, seed=42)
df_3 = C_df_model.sample(fraction=0.6, seed=42)
df_5 = C_df_model

df_1.cache()
df_3.cache()
df_5.cache()

print(df_1.count())
print(df_3.count())
print(df_5.count())

```

```python
import time

def measure_training_time(df, model):
    start = time.time()
    model.fit(df)
    end = time.time()
    return end - start
```

```python
time_1  = measure_training_time(df_1, rf)
time_3 = measure_training_time(df_3, rf)
time_5   = measure_training_time(df_5, rf)

print("Small:", time_1)
print("Medium:", time_3)
print("Full:", time_5)
```

```python
# for centralized data set 
# the time for computing all levels of data is almost same even for small and medium 
# scalability is less compared to distributed data 
```
