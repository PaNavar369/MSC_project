```python
from pyspark.sql import SparkSession
spark=SparkSession.builder\
      .appName("Thesis pyspark transformations")\
      .getOrCreate()
print("here")
spark
```

```python

df=spark.read.csv("/home/ubuntu/thesis/data/raw_data/train_1.csv", header=True, inferSchema=True)
df.show(5)
df.printSchema()
print("Total records in the data frame")
df.count()
#df.describe().show()
```

```python
#Pyspark transformations(unpivoting)

from pyspark.sql.functions import expr
columns =df.columns
date_columns=df.columns[1:]
print(columns)
#print(columns[1:])
counting_rows=df.count()
#print(counting_rows)

distinct_rows= df.distinct().count()
#print(distinct_rows)

# stacking 
from pyspark.sql.functions import col, explode, array, struct, lit

pairs = array(*[
    struct(
        lit(c).alias("Date"),
        col(c).alias("page_count")
    )
    for c in date_columns
])

stacked_df = df.select("Page", explode(pairs).alias("tmp")) \
               .select("Page", "tmp.Date", "tmp.page_count")

stacked_df.show(5)
```

```python

print(stacked_df.head(10))
stacked_df.show(5)
stacked_df.count()
```

```python
from pyspark.sql.functions import isnan,when,count
stacked_df.select([count(when(isnan(c) | col(c).isNull() ,c)).alias(c) for c in stacked_df.columns]).show()
```

```python
stacked_df_HNA=stacked_df.fillna({"page_count":0})
```

```python
from pyspark.sql.functions import isnan,when,count
stacked_df_HNA.select([count(when(isnan(c) | col(c).isNull(),c)).alias(c) for c in stacked_df_HNA.columns]).show()
```

```python
stacked_df_HNA.count()
```

```python
stacked_df_HNA.printSchema()
from pyspark.sql.functions import to_date
stacked_df_HNA=stacked_df_HNA.withColumn("Date",to_date("Date"))
stacked_df_HNA.printSchema()
stacked_df_HNA=stacked_df_HNA.orderBy("page","Date")
stacked_df_HNA=stacked_df_HNA.dropDuplicates(["page","Date"]) #Total duplicates i got is 10
stacked_df_HNA.count()
```

```python
processed_df= stacked_df_HNA
```

```python
processed_df.show(5)
processed_df.count()
```

```python
processed_df.write.mode("overwrite").parquet("/home/ubuntu/thesis/data/processed_data")
```
